# Personal auto claims EDA

Explore reporting volume, coverage, payout mix, reserve changes, development and vendor patterns.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display

root = Path.cwd()
if not (root / "deployment").exists():
    root = root.parent
snapshot = root / "deployment/snapshot"
profile = json.loads((snapshot / "data_report.json").read_text())
model = json.loads((snapshot / "model.json").read_text())
def sql(name):
    return pd.read_csv(snapshot / "analytics" / f"{name}.csv")
```

```python
frequency = sql("frequency")
display(frequency.sort_values("claims",ascending=False).head())
display(sql("coverage"))
display(sql("settlement"))
display(sql("supplements").tail())
```

```python
display(sql("paid_losses").pivot_table(index="category",values="dollars",aggfunc="sum"))
display(sql("reserves"))
display(sql("cohorts").tail())
display(sql("vendors").groupby("specialization")[["mean_paid","days","supplement_rate"]].mean())
```

Severity and repair specialization shift the mix of paid costs. Calendar paid losses aggregate amounts by payment date, while cohort development anchors amounts to first notice. Unadjusted vendor costs are useful for planning and can reflect the mix of claims sent to each shop.
